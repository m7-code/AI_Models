```python
# ============================================================
# PASTE THIS AS A NEW CELL IN COLAB — run it BEFORE the ngrok cell.
# This starts a real API server on port 8000 that your backend
# can call, instead of the manual upload()/input() cells.
# ============================================================

!pip install fastapi uvicorn python-multipart nest_asyncio -q

import os
import glob
import shutil
import subprocess
import nest_asyncio
import uvicorn
from fastapi import FastAPI, UploadFile, File
from fastapi.responses import FileResponse
from fastapi.middleware.cors import CORSMiddleware

nest_asyncio.apply()

app = FastAPI()
app.add_middleware(
    CORSMiddleware,
    allow_origins=["*"],
    allow_methods=["*"],
    allow_headers=["*"],
)

UPLOAD_DIR = "/content/SadTalker/api_uploads"
os.makedirs(UPLOAD_DIR, exist_ok=True)


@app.post("/generate-talking-video")
async def generate_talking_video(
    image: UploadFile = File(...),
    audio: UploadFile = File(...),
):
    # Save incoming files
    image_path = os.path.join(UPLOAD_DIR, image.filename)
    audio_path = os.path.join(UPLOAD_DIR, audio.filename)

    with open(image_path, "wb") as f:
        shutil.copyfileobj(image.file, f)
    with open(audio_path, "wb") as f:
        shutil.copyfileobj(audio.file, f)

    # Run SadTalker inference (same command you tested manually)
    result_dir = "/content/SadTalker/results"
    subprocess.run(
        [
            "python", "inference.py",
            "--driven_audio", audio_path,
            "--source_image", image_path,
            "--result_dir", result_dir,
            "--still",
            "--preprocess", "full",
        ],
        cwd="/content/SadTalker",
        check=True,
    )

    # Find the most recently created video
    videos = glob.glob(f"{result_dir}/*.mp4") + glob.glob(f"{result_dir}/**/*.mp4")
    latest_video = max(videos, key=os.path.getctime)

    return FileResponse(latest_video, media_type="video/mp4", filename="talking_avatar.mp4")


@app.get("/health")
def health():
    return {"status": "ok"}


# Run the server in a background thread (avoids Colab's asyncio event-loop conflict)
import threading

def run_server():
    uvicorn.run(app, host="0.0.0.0", port=8000)

server_thread = threading.Thread(target=run_server, daemon=True)
server_thread.start()

print("Server starting in background thread on port 8000...")
import time
time.sleep(3)
print("Should be up now — test with the ngrok URL + /health")
```
